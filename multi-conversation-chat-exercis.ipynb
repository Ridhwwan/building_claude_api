```python
#Load env variables

from dotenv import load_dotenv

load_dotenv()
```

```python
#Create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-5"
```

```python
#to store user messages
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

#to store assistant messages
def add_assistant_message(message, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

#to send messages to the API and get contextual responses
def chat(messages):
    message = client.messages.create(
        model="model",
        max_tokens=1000,
        messages=messages
    )
    return message.content[0].text
```

```python
# Make the initial list of messages
messages = []

# Use a 'while True' loop to run the chatbot forever
while True:
    # Get user input
    user_input = input("You: ")
    print("You:", user_input)

    # Add user input to the list of messages
    add_user_message(messages, user_input)

    # Call Claude with the 'chat' function
    answer = chat(messages)

    # Add generated text to the list of messages
    add_assistant_message(messages, answer)

    # Print the generated text
    print("Claude:", answer)
```
