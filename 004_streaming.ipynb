````python
#Create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-6"
````

````python
#to store user messages
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

#to store assistant messages
def add_assistant_message(message, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

#to send messages to the API and get contextual responses
def chat(messages, system=None, temperature=0.5, stop_sequences=None):
    params = {
        "model" : model,
        "max_tokens" : 1000,
        "messages" : messages,
        "temperature" : temperature
    }

    if system:
        """
        If a system prompt was passed in. then add it as the systems key inside
        the params dictionary. This will be used to set the system prompt for the chat.
        """
        params["system"] = system

    if temperature:
        params["temperature"]=temperature
        """
        If a temperature value was passed in. then add it as the systems key inside
        the params dictionary. This will be used to set the system prompt for the chat.
        """
    if stop_sequences:
        params["stop_sequences"] = stop_sequences


    # a variable with a multi-line string of system prompt
    system = """
    You are a patient math teacherYou want students to understand 
    the basics of mathematics before moving on to more complex topics. 
    Guide them to understand the origins of the concepts so they 
    understand how it applies in different ways moving forward.
    """

    message = client.messages.create(**params)
    return message.content[0].text
````

````python
# Make a list of messages
messages = []

# add in a user message to the list of messages using the add_user_message function
add_user_message(messages, "Generate a one sentence movie idea in Yourba language")
# I'll then get an answer by creating an "answer" variable that calls the "chat" function and passes in the "messages" list

try:
    answer = chat(messages, temperature=0.7)
except Exception as e:
    answer = (
        "Anthropic API error: "
        f"{e}. "
        "This usually means your API key is missing/invalid or your Anthropic account has no available credits. "
        "Add a valid API key and/or purchase credits in Anthropic billing."
    )

answer
````
